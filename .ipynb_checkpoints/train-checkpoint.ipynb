```python
import os
import pandas as pd
import librosa
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping
import tensorflow as tf
import joblib
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

CSV_PATH = "emotions-on-audio-dataset/versions/2/speech_emotions.csv"
AUDIO_BASE = "emotions-on-audio-dataset/versions/2/files"

df = pd.read_csv(CSV_PATH)

emotion_map = {
    "euphoric": 0,
    "joyfully": 1,
    "sad": 2,
    "surprised": 3
}

def extract_features_from_array(audio, sr=22050):
    """Extract features directly from an audio time-series array."""
    try:
        # Ensure consistent length (3 seconds)
        target_length = sr * 3
        if len(audio) < target_length:
            audio = np.pad(audio, (0, target_length - len(audio)), mode='constant')
        else:
            audio = audio[:target_length]

        features = []

        # MFCC features
        mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=20, n_fft=2048, hop_length=512)
        features.extend(np.mean(mfcc, axis=1))
        features.extend(np.std(mfcc, axis=1))

        # Spectral features
        spectral_centroid = librosa.feature.spectral_centroid(y=audio, sr=sr)
        features.append(np.mean(spectral_centroid))
        features.append(np.std(spectral_centroid))

        spectral_rolloff = librosa.feature.spectral_rolloff(y=audio, sr=sr)
        features.append(np.mean(spectral_rolloff))
        features.append(np.std(spectral_rolloff))

        # Zero crossing rate
        zcr = librosa.feature.zero_crossing_rate(audio)
        features.append(np.mean(zcr))
        features.append(np.std(zcr))

        # RMS energy
        rms = librosa.feature.rms(y=audio)
        features.append(np.mean(rms))
        features.append(np.std(rms))

        # Chroma features
        chroma = librosa.feature.chroma_stft(y=audio, sr=sr)
        features.extend(np.mean(chroma, axis=1))

        # Spectral contrast
        contrast = librosa.feature.spectral_contrast(y=audio, sr=sr)
        features.extend(np.mean(contrast, axis=1))

        return np.array(features)
    except Exception as e:
        return None

def extract_robust_features(file_path):
    """Load audio from file and extract features."""
    try:
        audio, sr = librosa.load(file_path, sr=22050, duration=3.0)
        return extract_features_from_array(audio, sr)
    except Exception as e:
        print(f"Error loading {file_path}: {str(e)}")
        return None

def augment_audio(audio, sr, aug_type):
    """Apply audio augmentation on time-series array."""
    if aug_type == 0:
        # Time stretching
        rate = np.random.uniform(0.9, 1.1)
        return librosa.effects.time_stretch(audio, rate=rate)
    elif aug_type == 1:
        # Pitch shifting
        steps = np.random.randint(-1, 2)
        return librosa.effects.pitch_shift(audio, sr=sr, n_steps=steps)
    else:
        # Add noise
        noise = np.random.normal(0, 0.001, audio.shape)
        return audio + noise

print("1. Collecting base audio file paths and labels...")
base_samples = []
base_labels = []

for idx, row in df.iterrows():
    speaker_id = row["set_id"]
    speaker_folder = os.path.join(AUDIO_BASE, str(speaker_id))

    for emotion, label in emotion_map.items():
        audio_path = os.path.join(speaker_folder, emotion + ".wav")
        if os.path.exists(audio_path):
            base_samples.append(audio_path)
            base_labels.append(label)

base_samples = np.array(base_samples)
base_labels = np.array(base_labels)

print(f"Total base audio files found: {len(base_samples)}")

# STEP 1: SPLIT FIRST to prevent data leakage
train_files, test_files, y_train_base, y_test_raw = train_test_split(
    base_samples, base_labels,
    test_size=0.2,
    random_state=42,
    stratify=base_labels,
    shuffle=True
)

print(f"Base Training files: {len(train_files)}, Base Test files: {len(test_files)}")

# STEP 2: Process Test Set (Unaugmented clean test data)
print("\n2. Processing Test Set (No Augmentation)...")
X_test_raw = []
y_test_list = []

for path, label in zip(test_files, y_test_raw):
    feats = extract_robust_features(path)
    if feats is not None:
        X_test_raw.append(feats)
        y_test_list.append(label)

X_test_raw = np.array(X_test_raw)
y_test_ml = np.array(y_test_list)

# STEP 3: Process Training Set & Apply Augmentation ONLY to Training Data
print("3. Processing Training Set (With Augmentations)...")
X_train_raw = []
y_train_list = []

for path, label in zip(train_files, y_train_base):
    try:
        audio, sr = librosa.load(path, sr=22050, duration=3.0)
        
        # Original features
        orig_feats = extract_features_from_array(audio, sr)
        if orig_feats is not None:
            X_train_raw.append(orig_feats)
            y_train_list.append(label)

        # 3 Augmented versions generated directly from array
        for aug_type in range(3):
            aug_audio = augment_audio(audio, sr, aug_type)
            aug_feats = extract_features_from_array(aug_audio, sr)
            if aug_feats is not None:
                X_train_raw.append(aug_feats)
                y_train_list.append(label)
    except Exception as e:
        continue

X_train_raw = np.array(X_train_raw)
y_train_ml = np.array(y_train_list)

print(f"Final Augmented Training Set shape: {X_train_raw.shape}")
print(f"Final Test Set shape: {X_test_raw.shape}")

# STEP 4: Scaling (Fit ONLY on training set to prevent data leakage)
scaler = StandardScaler()
X_train_ml = scaler.fit_transform(X_train_raw)
X_test_ml = scaler.transform(X_test_raw)

# Categorical targets for Neural Network
y_train_nn = to_categorical(y_train_ml, num_classes=4)
y_test_nn = to_categorical(y_test_ml, num_classes=4)

# STEP 5: Model Training
# Neural Network
def create_emotion_model(input_dim):
    return Sequential([
        Dense(128, activation='relu', input_shape=(input_dim,)),
        BatchNormalization(),
        Dropout(0.3),
        Dense(64, activation='relu'),
        BatchNormalization(),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dropout(0.2),
        Dense(4, activation='softmax')
    ])

input_dim = X_train_ml.shape[1]
nn_model = create_emotion_model(input_dim)
nn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    ReduceLROnPlateau(monitor='val_accuracy', factor=0.5, patience=15, min_lr=1e-7, verbose=0),
    EarlyStopping(monitor='val_accuracy', patience=30, restore_best_weights=True, verbose=0)
]

print("\nTraining Neural Network...")
nn_history = nn_model.fit(
    X_train_ml, y_train_nn,
    epochs=200,
    batch_size=8,
    validation_split=0.15,
    callbacks=callbacks,
    verbose=0
)

nn_test_loss, nn_test_accuracy = nn_model.evaluate(X_test_ml, y_test_nn, verbose=0)

# Traditional ML Models with regularized hyper-parameters to prevent overfitting
print("Training Traditional ML Models...")
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=5,
    random_state=42
)
rf_model.fit(X_train_ml, y_train_ml)
rf_pred = rf_model.predict(X_test_ml)
rf_accuracy = accuracy_score(y_test_ml, rf_pred)

svm_model = SVC(kernel='rbf', C=1.0, random_state=42, probability=True)
svm_model.fit(X_train_ml, y_train_ml)
svm_pred = svm_model.predict(X_test_ml)
svm_accuracy = accuracy_score(y_test_ml, svm_pred)

print(f"Neural Network Accuracy: {nn_test_accuracy:.4f}")
print(f"Random Forest Accuracy:  {rf_accuracy:.4f}")
print(f"SVM Accuracy:            {svm_accuracy:.4f}")

# Save Best Model
if rf_accuracy >= nn_test_accuracy and rf_accuracy >= svm_accuracy:
    joblib.dump(rf_model, 'best_emotion_model.pkl')
    best_model_type, best_accuracy, best_pred = "Random Forest", rf_accuracy, rf_pred
elif nn_test_accuracy >= rf_accuracy and nn_test_accuracy >= svm_accuracy:
    nn_model.save("best_emotion_model.h5")
    best_model_type = "Neural Network"
    best_accuracy = nn_test_accuracy
    best_pred = np.argmax(nn_model.predict(X_test_ml), axis=1)
else:
    joblib.dump(svm_model, 'best_emotion_model.pkl')
    best_model_type, best_accuracy, best_pred = "SVM", svm_accuracy, svm_pred

joblib.dump(scaler, 'feature_scaler.pkl')

print(f"\n=== FINAL RESULTS ===")
print(f"Best Model: {best_model_type}")
print(f"Realistic Best Accuracy: {best_accuracy:.4f}")

print("\nClassification Report:")
print(classification_report(y_test_ml, best_pred, target_names=list(emotion_map.keys())))

print("Confusion Matrix:")
print(confusion_matrix(y_test_ml, best_pred))
```
